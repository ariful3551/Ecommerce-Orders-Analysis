##### Checking Calculated Columns
*Checking whether `Sales` and `Profit` follow a formula based on the other columns.*

```python
rel_err = (df["Sales"] - df["Quantity"] * df["Unit_Price"] * (1 - df["Discount"])).abs() / df["Sales"]
print("Sales figures match (within a 1% margin of error):", (rel_err < 0.01).mean().round(3))

ok = df[df["Quantity"] > 0]
print("Profit = Sales - Quantity*Cost:",
      ((ok["Profit"] - (ok["Sales"] - ok["Quantity"] * ok["Cost"])).abs() < 0.05).mean().round(3))
```

```python
no_match = df[~(rel_err < 0.01)]
print("  Rows not matching:", len(no_match))
print("  Discount is missing:", no_match["Discount"].isna().sum())
print("  Quantity <= 0:", (no_match["Quantity"] <= 0).sum())
print("  Discount outside 0-0.9:", ((no_match["Discount"] < 0) | (no_match["Discount"] > 0.9)).sum())
```

**Observation:**
- In 96% of cases (4,817 out of 5,020 rows)—allowing for a 1% margin of error—the `Sales` value follows the formula: `Quantity * Unit_Price * (1 - Discount)`.
- The 203 rows that do not conform to this formula contain errors in the input data: 199 rows have a missing `Discount` value, 3 rows have a `Quantity` of 0 or less, and 2 rows contain an invalid `Discount` value. When all input data is correct, the formula holds true for every single row.
- The equation `Profit = Sales - Quantity * Cost` holds true for every row where the quantity is positive; here, `Cost` refers to the **cost per unit**, not the total cost.
- Since the `Sales` value is consistent with the formula, it may be possible to recover a missing `Discount` value in any given row by using the information available in the other columns. This will be examined in the second step (Step 2).
