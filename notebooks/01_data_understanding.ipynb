##### Missing Values
*Counting how many values are missing in each column.*

```python
missing = df.isna().sum().to_frame("missing")
missing["pct"] = (missing["missing"] / len(df) * 100).round(1)
missing[missing["missing"] > 0].sort_values("missing", ascending=False)
```

```python
rows_missing = df.isna().any(axis=1).sum()
print(f"Total null row: {rows_missing} and it's {round(rows_missing / len(df) * 100, 1)} % of total row.")
```

**Observations:**

- There are missing values ​​in 10 columns. The remaining 18 columns are complete.
- The `Customer_Rating` column has the highest number of missing values ​​(283 rows, 5.6%), followed by `Delivery_Days` (202 rows, 4.0%) and `Discount` (199 rows, 4.0%).
- In the remaining 7 columns (`Gender`, `Age`, `Payment_Method`, `Marketing_Channel`, `Customer_Segment`, `City`, `Customer_Name`), the proportion of missing values ​​ranges from 2.4% to 3.5%.
- No column has more than 6% missing values; therefore, it is not necessary to drop any column entirely.
- The missing values ​​are distributed across several columns.If every row containing at least one missing value is deleted, approximately 29% of the data (1,451 out of 5,020 rows) will be lost; therefore, deleting rows is not a good solution. In the second step, a decision will be made for each column individually regarding how to handle the missing values.

### Categorical Columns: Value Counts
*Checking which values appear in each text column, to find inconsistent spellings.*

```python
text_cols = ["Gender", "Country", "Category", "Sub_Category", "Customer_Segment",
             "Payment_Method", "Order_Status", "Shipping_Method",
             "Marketing_Channel", "Device_Type"]
for c in text_cols:
    print(f"--- {c} ---")
    print(df[c].value_counts(dropna=False), "\n")
```

```python
for c in ["Gender", "Country", "Category"]:
    raw_n = df[c].nunique()
    clean_n = df[c].str.strip().str.lower().nunique()
    print(c, "| as written:", raw_n, "| after strip + lowercase:", clean_n)
```

**Observations:**
- **Gender:** There are 12 different spellings or variations (e.g., `m`, ` m `, `M`, `male`, `Male`, `MALE`, `f`, `F`, `female`, `Woman`, `woman`), and in some instances, the value is missing. Yet, these all refer to only two categories.
- **Country:** There are 23 distinct labels for just 6 countries (for example, Germany is represented as `DE`, `Germany`, `germany`, `Deutschland`, and even as `Germany `—with an extra space). For other countries, both local names (e.g., `Österreich`, `Schweiz`) and country codes (e.g., `NL`, `AT`, `CH`, `BE`, `FR`) have been used.
- **Category:** There are 17 distinct labels for just 5 categories (e.g., `Electronic`, `Electronics`, `electronics`, `ELEC`; or `Home & Kitchen`, `Home and Kitchen`, `home_kitchen`).
- **Sub_Category, Customer_Segment, Payment_Method, Order_Status, Shipping_Method, Marketing_Channel, Device_Type:** The values ​​in these columns are quite consistent, with only a small number of distinct labels for each.
- Before performing any grouping or calculations, these columns need to be standardized (mapped) in the second step.
